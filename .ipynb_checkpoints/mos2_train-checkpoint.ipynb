# Mask R-CNN: Inference and Testing on MoS2 Microscopy Images

  This Notebook is designed to load pre-trained Mask R-CNN model weights and perform instance segmentation inference on novel MoS2 microscopy images (bright-field and dark-field). 
  The code automatically identifies single-layer regions, multi-layer regions, and contaminants within the images. It generates visualization results—including bounding boxes and class labels—alongside binarized masks for subsequent quantitative analysis.

Environment Setup and Dependencies

```python
import os
# Block TensorFlow's underlying C warning messages (INFO and WARNING)
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

import sys
import numpy as np
import tensorflow as tf
import cv2
from skimage import io
import matplotlib.pyplot as plt

# Get the project root directory
ROOT_DIR = os.path.abspath("./")
# Add the project directory to the system path to import the mrcnn module
sys.path.append(ROOT_DIR)

from mrcnn import utils
from mrcnn import visualize
import mrcnn.model as modellib

print("TensorFlow Version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))
```

# 1. Path and Inference Configuration

In this section, we define the directory paths for logs and pre-trained model weights. We also initialize the `InferenceConfig` by overriding the base `CocoConfig` defined in `tdmcoco.py`. For the inference phase, the `BATCH_SIZE` is strictly set to 1 (calculated as `GPU_COUNT` × `IMAGES_PER_GPU`).

```python
# Log and Training Model Save Directory
MODEL_DIR = os.path.join(ROOT_DIR, "logs")

# Path to the trained local weight file
COCO_MODEL_PATH = os.path.join(ROOT_DIR, "mask_rcnn_tdm_0400.h5")

# Import your custom dataset configuration
import mos2_train
base_config = mos2_train.CocoConfig()

# Define inference configuration
class InferenceConfig(base_config.__class__):
# Each GPU processes 1 image during inference
    GPU_COUNT = 1
    IMAGES_PER_GPU = 1
    
# Number of categories: 1 (Background)   3 (Single layer, Contaminants, Multiple layers)
    NUM_CLASSES = 1 + 3
    
# Set a minimum confidence threshold to filter out low-scoring predicted boxes
    DETECTION_MIN_CONFIDENCE = 0.7

config = InferenceConfig()
config.display()
```

# 2. Model Initialization

In this step, the Mask R-CNN architecture is instantiated in `inference` mode. Subsequently, the pre-trained weights are loaded from the designated `.h5` file to prepare the model for object detection and segmentation tasks.

```python
model = modellib.MaskRCNN(mode="inference", model_dir=MODEL_DIR, config=config)

# Load weights
print("Loading weights from:", COCO_MODEL_PATH)
model.load_weights(COCO_MODEL_PATH, by_name=True)
print("Weights loaded successfully!")
```

# 3. Image Inference and Result Export

In this section, we define the `predict_and_save` function to automate the batch processing of MoS2 micrographs. The function iterates through the input directory, feeds each image into the Mask R-CNN model for inference, and invokes `visualize.display_instances` to overlay segmentation masks and categorical labels. Finally, both the annotated visualizations and the standalone binary masks are exported to designated folders for downstream quantitative analysis.

```python
def predict_and_save(input_dir, output_dir, mask_dir, mask2_dir):
    # Ensure the output directory exists
    os.makedirs(output_dir, exist_ok=True)
    os.makedirs(mask_dir, exist_ok=True)
    os.makedirs(mask2_dir, exist_ok=True)
    
    valid_extensions = ['.jpg', '.jpeg', '.png']
    class_names = ['', 'single layer', 'contaminants', 'multi layer']

    for filename in os.listdir(input_dir):
        if any(filename.lower().endswith(ext) for ext in valid_extensions):
            image_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, f"result_{filename}")
            mask_path = os.path.join(mask_dir, f"mask_{os.path.splitext(filename)[0]}.png")
            mask2_path = os.path.join(mask2_dir, f"mask_{os.path.splitext(filename)[0]}.png")
            
            # Load image
            image = io.imread(image_path)
            orig_shape = image.shape
            
            # Perform object detection
            results = model.detect([image], verbose=0)
            r = results[0]
            
            # Obtain OpenCV visualizations and two types of masks
            result_image, binary_mask, mask2 = visualize.display_instances(
                image, r['rois'], r['masks'], r['class_ids'],
                class_names, r['scores']
            )
            
            # Resize to original dimensions and save
            result_image = cv2.resize(result_image, (orig_shape[1], orig_shape[0]))
            cv2.imwrite(output_path, cv2.cvtColor(result_image, cv2.COLOR_RGB2BGR))
            
            binary_mask = cv2.resize(binary_mask, (orig_shape[1], orig_shape[0]))
            cv2.imwrite(mask_path, binary_mask)
            
            binary2_mask = cv2.resize(mask2, (orig_shape[1], orig_shape[0]))
            cv2.imwrite(mask2_path, binary2_mask)

            print(f"Processed and Saved: {filename}")
```

# 4. Executing Batch Processing

Define your input and output directory paths here. Executing the code block below will automatically process all micrographs residing in the `Bimage` folder using the inference pipeline.

```python
# Path Configuration
INPUT_FOLDER = os.path.join(ROOT_DIR, "Bimage")
OUTPUT_FOLDER = os.path.join(ROOT_DIR, "result")
MASK_FOLDER = os.path.join(ROOT_DIR, "masks")
MASK_FOLDER2 = os.path.join(ROOT_DIR, "masks2")

print(f"Starting prediction on images in: {INPUT_FOLDER}")
predict_and_save(INPUT_FOLDER, OUTPUT_FOLDER, MASK_FOLDER, MASK_FOLDER2)
print("All tasks completed!")
```

# 5. Visualize a Single Image

```python
%matplotlib inline
# Randomly select or specify a test image
sample_image_name = "2.jpeg"  # Replace with an image filename from the Bimage folder
sample_path = os.path.join(INPUT_FOLDER, sample_image_name)

if os.path.exists(sample_path):
    image = io.imread(sample_path)
    results = model.detect([image], verbose=1)
    r = results[0]
    
    class_names = ['', 'single layer', 'contaminants', 'multi layer']

    result_image, mask1, mask2 = visualize.display_instances(
        image, r['rois'], r['masks'], r['class_ids'], 
        class_names, r['scores']
    )
    
    fig, axes = plt.subplots(1, 3, figsize=(20, 8)) 
     
    axes[0].imshow(result_image)
    axes[0].set_title(f"Prediction Result: {sample_image_name}", fontsize=16)
    axes[0].axis('off')
    
    axes[1].imshow(mask1, cmap='gray') 
    axes[1].set_title("All Masks (Binary)", fontsize=16)
    axes[1].axis('off')
    
    axes[2].imshow(mask2, cmap='gray')
    axes[2].set_title("multi layer Mask", fontsize=16)
    axes[2].axis('off')
    
    plt.tight_layout()
    plt.show()        
```
